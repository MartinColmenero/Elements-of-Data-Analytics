### **Question One:** (2 points) Select all rows where the carrier is 'UA' and the difference between departure time and scheduled departure time is more than 2 minutes.


### **Question Two:** (2 points) How many unique airlines are there in the dataset?

```python
#TODO: Your code goes here
unique_airlines = df.groupby('carrier').size()
unique_airlines.size

## (14)
## There are 14 unique airlines
```

### **Question Three:** (3 points) Which carrier has the longest cumulative departure delay?

```python
#TODO: Your code goes here
# Group by carrier and sum the departure delays, then sort in descending order
# Get the carrier with the longest delays in sum
carrier_delay_list = df.groupby('carrier')['dep_delay'].sum()
carrier_delay_list.sort_values(ascending = False) # Displays list in descending order

carrier_with_longest_delay = carrier_delay_list['EV'] # 'EV' had the longest departure delays
print(carrier_with_longest_delay)
```

 ### **Question Four:** (3 points) Which carrier has the single longest departure delay?
